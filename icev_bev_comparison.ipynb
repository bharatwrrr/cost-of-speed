```python
import os
import fireducks.pandas as pd
import numpy as np
```

```python
folder_path = '/path/to/overspeeding_analysis/figure_data/'
summary_dict = {}
for DAY in [1,2,3,4]:
    file_name = f'per_class_veh_os_ca_bev_day{DAY}.csv'
    veh_class_os_df = pd.read_csv(os.path.join(folder_path, file_name), index_col=0)
    for veh_class in veh_class_os_df.index:
        if veh_class not in summary_dict:
            summary_dict[veh_class] = {}
            summary_dict[veh_class]['avg_max_os'] = [veh_class_os_df.loc[veh_class, 'avg_max_os']]
            summary_dict[veh_class]['avg_dist'] = [veh_class_os_df.loc[veh_class, 'dist_mi'] / 
                                                  veh_class_os_df.loc[veh_class, 'num_trips']]
            summary_dict[veh_class]['num_trips'] = [veh_class_os_df.loc[veh_class, 'num_trips']]
            summary_dict[veh_class]['num_os_trips'] = [veh_class_os_df.loc[veh_class, 'num_os_trips']]
            summary_dict[veh_class]['avg_time_os'] = [veh_class_os_df.loc[veh_class, 'avg_time_os']]
            summary_dict[veh_class]['total_time_os'] = veh_class_os_df.loc[veh_class, 'total_time_os']
            summary_dict[veh_class]['total_trip_time'] = veh_class_os_df.loc[veh_class, 'total_trip_time']
        else:
            summary_dict[veh_class]['avg_max_os'] += [veh_class_os_df.loc[veh_class, 'avg_max_os']]
            summary_dict[veh_class]['avg_dist'] += [veh_class_os_df.loc[veh_class, 'dist_mi'] / 
                                                  veh_class_os_df.loc[veh_class, 'num_trips']]
            summary_dict[veh_class]['num_trips'] += [veh_class_os_df.loc[veh_class, 'num_trips']]
            summary_dict[veh_class]['num_os_trips'] += [veh_class_os_df.loc[veh_class, 'num_os_trips']]
            summary_dict[veh_class]['avg_time_os'] += [veh_class_os_df.loc[veh_class, 'avg_time_os']]
            summary_dict[veh_class]['total_time_os'] += veh_class_os_df.loc[veh_class, 'total_time_os']
            summary_dict[veh_class]['total_trip_time'] += veh_class_os_df.loc[veh_class, 'total_trip_time']
            
```

```python
summary_df = pd.DataFrame(index=summary_dict.keys())
for veh_class in summary_dict:
    num_trips = np.array(summary_dict[veh_class]['num_trips'])
    total_trips = num_trips.sum()
    num_os_trips = np.array(summary_dict[veh_class]['num_os_trips'])
    avg_max_os = np.array(summary_dict[veh_class]['avg_max_os'])
    summary_df.loc[veh_class, 'avg_max_os'] = (avg_max_os * num_trips).sum() / total_trips
    summary_df.loc[veh_class, 'perc_trip_os'] = num_os_trips.sum() / total_trips * 100
    summary_df.loc[veh_class, 'perc_trip_time_os'] = summary_dict[veh_class]['total_time_os'] / summary_dict[veh_class]['total_trip_time'] * 100
    summary_df.loc[veh_class, 'avg_dist_mi'] = (np.array(summary_dict[veh_class]['avg_dist']) * num_trips).sum() / total_trips
    
```

```python
summary_df.to_csv('./data/')
```
